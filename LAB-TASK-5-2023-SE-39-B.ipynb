```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# --- IMPORTANT: Change this to the path of your image file ---
image_path = '/content/image5(a).png'

try:
    # Open the image file
    img = Image.open(image_path).convert('L')  # Convert to grayscale

    # Display the original image
    plt.imshow(img, cmap='gray')
    plt.axis('off')  # Hide axes for cleaner display
    plt.title(f'Original Image: {image_path.split("/")[-1]}')
    plt.show()

    # Convert the image to a NumPy array
    img_array = np.array(img)

    # Define the mean filter (kernel)
    kernel_size = 3
    mean_filter = np.ones((kernel_size, kernel_size), np.float32) / (kernel_size * kernel_size)

    # Apply the mean filter using convolution
    def apply_mean_filter(image_array, kernel):
        from scipy.ndimage import convolve
        return convolve(image_array, kernel, mode='reflect')

    # Apply the mean filter to the image
    filtered_image = apply_mean_filter(img_array, mean_filter)

    # Display the filtered image
    plt.imshow(filtered_image, cmap='gray')
    plt.axis('off')
    plt.title('Mean Filtered Image')
    plt.show()

except FileNotFoundError:
    print(f"Error: The image file was not found at '{image_path}'.")
    print("Please make sure the path is correct and the file exists.")
except Exception as e:
    print(f"An error occurred while trying to open or display the image: {e}")
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import median_filter

# Function to apply Median Filter
def apply_median_filter(image_path):
    try:
        # Open the image file
        img = Image.open(image_path).convert('L')  # Convert to grayscale
        img_array = np.array(img)

        # Display the original image
        plt.imshow(img_array, cmap='gray')
        plt.axis('off')
        plt.title('Original Image')
        plt.show()

        # Apply the Median filter
        filtered_image = median_filter(img_array, size=3)

        # Display the filtered image
        plt.imshow(filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Median Filtered Image')
        plt.show()

    except FileNotFoundError:
        print(f"Error: The image file was not found at '{image_path}'.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
image_path = '/content/image5(a).png'  # Change path as needed
apply_median_filter(image_path)
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import generic_filter

# Function to apply Mode Filter
def mode(arr):
    """Return the mode of the array."""
    values, counts = np.unique(arr, return_counts=True)
    return values[np.argmax(counts)]

# Function for applying Mode Filter
def apply_mode_filter(image_path):
    try:
        # Open the image file
        img = Image.open(image_path).convert('L')  # Convert to grayscale
        img_array = np.array(img)

        # Display the original image
        plt.imshow(img_array, cmap='gray')
        plt.axis('off')
        plt.title('Original Image')
        plt.show()

        # Apply the Mode filter
        filtered_image = generic_filter(img_array, mode, size=3)

        # Display the filtered image
        plt.imshow(filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Mode Filtered Image')
        plt.show()

    except FileNotFoundError:
        print(f"Error: The image file was not found at '{image_path}'.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
image_path = '/content/image5(a).png'  # Change path as needed
apply_mode_filter(image_path)
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# Function to apply Gaussian Filter
def apply_gaussian_filter(image_path):
    try:
        # Open the image file
        img = Image.open(image_path).convert('L')  # Convert to grayscale
        img_array = np.array(img)

        # Display the original image
        plt.imshow(img_array, cmap='gray')
        plt.axis('off')
        plt.title('Original Image')
        plt.show()

        # Apply the Gaussian filter
        filtered_image = gaussian_filter(img_array, sigma=1)  # Change sigma for different effects

        # Display the filtered image
        plt.imshow(filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Gaussian Filtered Image')
        plt.show()

    except FileNotFoundError:
        print(f"Error: The image file was not found at '{image_path}'.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
image_path = '/content/image5(a).png'  # Change path as needed
apply_gaussian_filter(image_path)
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import median_filter, generic_filter, gaussian_filter

# Function to apply filters and display results
def apply_filters_and_display(image_path):
    try:
        # Open the image file and convert to grayscale
        img = Image.open(image_path).convert('L')
        img_array = np.array(img)

        # Original Image
        original_image = img_array

        # Applying Median Filter
        median_filtered_image = median_filter(original_image, size=3)

        # Applying Mode Filter
        def mode(arr):
            """Return the mode of the array."""
            values, counts = np.unique(arr, return_counts=True)
            return values[np.argmax(counts)]
        mode_filtered_image = generic_filter(original_image, mode, size=3)

        # Applying Gaussian Filter
        gaussian_filtered_image = gaussian_filter(original_image, sigma=1)

        # Display all images together
        plt.figure(figsize=(15, 10))

        plt.subplot(2, 2, 1)
        plt.imshow(original_image, cmap='gray')
        plt.axis('off')
        plt.title('Original Image')

        plt.subplot(2, 2, 2)
        plt.imshow(median_filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Median Filtered Image')

        plt.subplot(2, 2, 3)
        plt.imshow(mode_filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Mode Filtered Image')

        plt.subplot(2, 2, 4)
        plt.imshow(gaussian_filtered_image, cmap='gray')
        plt.axis('off')
        plt.title('Gaussian Filtered Image')

        plt.tight_layout()
        plt.show()

    except FileNotFoundError:
        print(f"Error: The image file was not found at '{image_path}'.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
image_path = '/content/image5(a).png'  # Change path as needed
apply_filters_and_display(image_path)
```

1. Median Filter
Operation: Replaces each pixel with the median value of the neighboring pixels.
Effect on Original Image:
Effectively removes "salt and pepper" noise (random occurrences of bright and dark pixels).
Preserves edges better than linear filters; edges are less blurred.
Use Case: Best for images with impulsive noise.
2. Mode Filter
Operation: Replaces each pixel with the most frequently occurring pixel value in its neighborhood.
Effect on Original Image:
Similar to the median filter but focuses on the mode (most common value) instead.
Useful for images with consistent but varying noise.
Generally less effective than the median filter for noise reduction.
Use Case: Can be effective in removing noise while maintaining the characteristic features of an image.
3. Gaussian Filter
Operation: Applies a convolution with a Gaussian function (bell curve) to smooth the image.
Effect on Original Image:
Reduces high-frequency noise and blurs the image, which can result in a loss of detail.
The amount of blurring is adjustable through the sigma value (standard deviation of the Gaussian).
Use Case: Commonly used for image smoothing, reducing noise, and as a pre-processing step for other image processing tasks.
Summary of Differences
Feature	Median Filter	Mode Filter	Gaussian Filter
Pixel Replacement	Median of neighbors	Mode of neighbors	Weighted average (Gaussian distribution)
Noise Type	Best for salt and pepper	Works on general noise, less effective for impulsive noise	Good for high-frequency noise
Edge Preservation	Good	Moderate	Moderate to Poor
Blurring	Minimal	Minimal	Can be substantial depending on sigma
Computational Complexity	Moderate	Moderate	Usually higher due to convolution
Visual Effect on the Original Image
Median Filter: The image retains sharp edges while effectively removing noise. By taking the median, this filter smooths out extreme values.

Mode Filter: Results in a smoother image compared to the original but can lead to higher noise levels than the median, as it may replace pixels with less representative values.

Gaussian Filter: Typically results in a smoother, more uniform appearance with reduced detail. Ideal for preparing images for further processing, but it can obscure fine details.

Conclusion
Selecting the right filter depends on the type of noise in the image and the desired outcome. For impulsive noise, the median filter is preferable, while the Gaussian filter is better for smoothing effects. Each filter serves different purposes and affects the image in unique ways.
